## 📚 Importación de librerías
En esta sección se importan las librerías necesarias para el análisis y la visualización de datos.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 📂 Carga del dataset

El dataset utilizado en este proyecto contiene información sobre canciones de Spotify, incluyendo popularidad, energía, danceability, tempo y género musical.

Fuente del dataset:
https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset

El dataset se carga en un DataFrame de pandas para comenzar el proceso de análisis exploratorio de datos.

```python
df = pd.read_csv("../data/dataset.csv")
df.head()
```

## 🔍 Exploración del dataset

Se analiza la estructura general del dataset para comprender las variables disponibles, los tipos de datos y la existencia de posibles valores nulos.

```python
df.info()
```

## 📊Distribución de popularidad

Se utiliza un histograma para visualizar cómo se distribuye la popularidad de las canciones dentro del dataset.

```python
df['popularity'].hist()
plt.title("Distribución de popularidad")
plt.xlabel("Popularidad")
plt.ylabel("Cantidad de canciones")
plt.show()
```

### 📈 Resumen estadístico

La mayoría de las canciones presentan niveles de popularidad medios o bajos, mientras que solo una pequeña parte alcanza valores de popularidad elevados. Esto indica que las canciones extremadamente populares representan una minoría dentro del dataset.

## 🎼Relación entre danceability y popularidad

Se analiza la relación entre el nivel de danceability de las canciones y su popularidad media utilizando un gráfico de barras.

```python
danceability_groups = df.groupby(pd.cut(df['danceability'], bins=10))['popularity'].mean()

danceability_groups.plot(kind='bar')

plt.title("Average Popularity by Danceability")
plt.xlabel("Danceability Level")
plt.ylabel("Average Popularity")

plt.show()
```

### 📈 Resumen estadístico

Las canciones con mayores niveles de danceability tienden a presentar valores de popularidad más elevados. Esto puede indicar que las canciones más bailables suelen tener una mejor recepción entre los usuarios de Spotify.

## 🧠Diccionario de variables

Se utiliza un diccionario para describir el significado de algunas variables importantes del dataset.

```python
variables_description = {
    'popularity': 'Nivel de popularidad de la canción',
    'danceability': 'Qué tan bailable es una canción',
    'energy': 'Nivel de intensidad y energía',
    'tempo': 'Velocidad de la canción'
}

print(variables_description)
```

## ⚙️ Clasificación de popularidad

Se crea una función sencilla para clasificar canciones según su nivel de popularidad.

```python
def popularity_level(popularity):
    if popularity >= 75:
        return "High"
    elif popularity >= 50:
        return "Medium"
    else:
        return "Low"
```

```python
df['popularity_category'] = df['popularity'].apply(popularity_level)

df[['popularity', 'popularity_category']].head()
```

## 🧠 Correlaciones entre variables

Se analiza la relación entre diferentes variables numéricas del dataset para identificar posibles patrones dentro de las características musicales.

```python
correlation = df[['popularity', 'danceability', 'energy', 'tempo']].corr()

print(correlation)
```
